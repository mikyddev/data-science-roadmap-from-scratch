```python
import duckdb

# 1. Conectar a um banco de dados local ou em memória
con = duckdb.connect()

# 2. Simular uma tabela de vendas com grande volume de dados diretamente via SQL
con.execute("""
    CREATE TABLE vendas AS
    SELECT
        range AS id_transacao,
        (range % 100) + 1 AS id_loja,
        (range % 1000) + 1 AS id_cliente,
        ROUND((RANDOM() * 500) + 10, 2) AS valor_compra,
        CURRENT_DATE - INTERVAL (CAST(RANDOM() * 365 AS INT)) DAY AS data_venda
    FROM range(1000000); -- Simula 1 milhão de registros de vendas
""")

# 3. Fazer uma extração analítica agregada (Agrupar vendas totais por loja)
resultado = con.execute("""
    SELECT
        id_loja,
        COUNT(id_transacao) AS total_vendas,
        ROUND(SUM(valor_compra), 2) AS faturamento_total
    FROM vendas
    GROUP BY id_loja
    ORDER BY faturamento_total DESC
    LIMIT 10;
""").df()

# 4. Exibir o resultado
print(resultado)
```
