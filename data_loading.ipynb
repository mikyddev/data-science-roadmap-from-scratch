```python
import pandas as pd

# Step 1: Carregar os dados de uma URL pública
url = 'https://raw.githubusercontent.com/mwaskom/seaborn-data/master/penguins.csv'
df = pd.read_csv(url)

# Step 2: Inspecionar o tamanho do dataset (linhas, colunas)
print(f"Total de linhas e colunas lidas: {df.shape}")

# Step 3: Fazer uma extração e filtragem
# Filtrar apenas os pinguins da espécie 'Adelie' com massa corporal maior que 4000g
pinguins_grandes = df[(df['species'] == 'Adelie') & (df['body_mass_g'] > 4000)]

# Step 4: Salvar o resultado filtrado em um novo arquivo CSV
pinguins_grandes.to_csv('pinguins_adelie_grandes.csv', index=False)

print("Exercício concluído! Arquivo gerado com sucesso.")
```
